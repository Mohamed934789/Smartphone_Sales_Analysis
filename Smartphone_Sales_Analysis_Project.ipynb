# Smartphone Sales — exploratory learning analysis

Uses the committed CSV. Missing profit is **not** treated as observed profit. A 35% margin is shown only as an explicit sensitivity scenario. Associations do not prove that specifications cause sales.

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("smartphone_sales_dataset.csv")
print("Rows:",len(df)); print(df.isna().sum())
df=df.dropna(subset=["Price_USD"]).copy()
df["Sales_Revenue"]=df["Sales_Revenue"].fillna(df["Price_USD"]*df["Quantity_Sold"])
df["profit_is_imputed"]=df["Profit"].isna()
df["profit_scenario_35pct"]=df["Profit"].fillna(df["Sales_Revenue"]*.35)
print("Observed profit rows:", (~df["profit_is_imputed"]).sum())
print("Scenario-imputed profit rows:",df["profit_is_imputed"].sum())
```

```python
print("Revenue:",df["Sales_Revenue"].sum())
print("Observed profit total (incomplete coverage):",df["Profit"].sum(min_count=1))
print("Profit under 35% imputation scenario:",df["profit_scenario_35pct"].sum())
df.groupby("Brand")["Quantity_Sold"].sum().sort_values().plot.barh(title="Units sold by brand")
plt.tight_layout(); plt.show()
```

```python
print(df.groupby("OS").agg(revenue=("Sales_Revenue","sum"),units=("Quantity_Sold","sum"),rating=("Rating","mean")))
print("RAM/price association with units:")
print(df[["RAM_GB","Price_USD","Battery_mAh","Quantity_Sold"]].corr()["Quantity_Sold"])
df.plot.scatter(x="Price_USD",y="Quantity_Sold",title="Price vs units: association only"); plt.show()
```

## Limits
The dataset origin and whether values are synthetic require owner confirmation. Profit conclusions depend on missing-data treatment. Revenue and profit overlap and must not be presented as disjoint pie slices. This is not a market forecasting model.
